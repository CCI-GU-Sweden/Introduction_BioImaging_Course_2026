# 2. Variables

A **variable** is a name that refers to a value.

We create a variable using `=`.


````python
# create a variable and print it
my_number = 10
print(my_number)
````

Read this as:

- put the value `"Alex"` into a variable called `name`
- put the value `24` into a variable called `age`

The variable name is on the **left** side of `=`.  
The value is on the **right** side.


````python
# create two variables and print them in one line
name = "Alex"
age = 24
print(name, age)
````

Variables can be changed:

````python
# create a numeric variable, print it, then change it and print it again
my_number = 11
print(my_number)

my_number = 5
print(my_number)
````

## Integers

Integers are whole numbers.


````python
# create two integer variables and subtract them
integer_1 = 5
integer_2 = 2
print(integer_1 - integer_2)
````

You can do typical mathematical operations with integers

````python
# create two variables, print the sum, difference, product, quotient and one raised to the power of the other

var_1 = 5
var_2 = 2

print(var_1 + var_2) # sum
print(var_2 - var_1) # difference
print(var_1 * var_2) # product
print(var_1 / var_2) # division which will give an int or a float
print(var_1 // var_2) # division which rounds down the quotient to an int
print(var_1 % var_2) # modulo: gives you the remainder after dividing
print(var_1 ** var_2) # exponent
````

# 5. Making decisions with `if`

Programs often need to make decisions.

An `if` statement runs code **only when a condition is true**.


````python
# create a variable, and print a message if it fulfils some condition
x = 1
if x > 3:
    print("Larger than 3")

print("Completed")
````

Two important syntax rules:

```python
if condition:
    indented_code
```

1. the `if` line ends with `:`
2. the code inside the `if` statement is indented


What happens if the condition is false, and after the if statement is complete?

````python
temperature = 25

if temperature > 20:
    print("It is warm.")

print("Finished checking the temperature.")
````

## `if` and `else`

Use `else` when you want one thing to happen if the condition is true and something else if it is false.


````python
# demonstrate if/else statements
x = 1
if x > 2:
    print("Larger")
else:
    print("Smaller")
````

## `if`, `elif`, and `else`

Use `elif` when there are several possible cases. The lines of code are executed in order, and if the condition is fulfiled, then we exit the if statement.


````python
# demonstrate if, elif and else. Print something after termination.
x = 3
if x > 4:
    print("Massive")
elif x > 2:
    print("Big")
else:
    print("Small")
````

Python checks the conditions from top to bottom and stops at the first condition that is true.

The term after the if statement is a boolean! ```If``` is checking whether this boolean is True or False

````python
# create a boolean variable from some condition, and use that in an if statement
x = 5
my_bool = x < 8
print(my_bool)
if my_bool:
    print("Smaller")
````

# 6. Lists

A **list** stores multiple values in one variable.

Lists use square brackets `[ ]`.


````python
# create a list of floats and print it
my_list = [1.5, 2.5, 3.5]
print(my_list)
````

Lists can contain strings:

````python
# create a list of strings and print it
my_list = ["hello", "good", "morning"]
my_list
````

We can access individual items using their position.

Python starts counting at **0**.


````python
# print each element of the list above
print(my_list[0])
print(my_list[1])
print(my_list[2])
````

````python
# counting starts at 0, so using index 1 means we retrieve the second element
item_1 = my_list[1]
print(item_1)
````

````python
# print the last element of a list
print(my_list[-1])
````

So:

| Position | Index | Value |
|---:|---:|---|
| first | `0` | `"Control"` |
| second | `1` | `"Treatment A"` |
| third | `2` | `"Treatment B"` |

This zero-based indexing is very common in Python.


We can check how many items are in a list with `len()`.

````python
# demonstrate len() of some list
print(1+1)
len(my_list)
````

# 7. `for` loops

A `for` loop repeats code.

This is especially useful when we have a list. When we use a for loop to access the terms in a list one by one, this is called **looping** or more formally, **iteration**.


````python
# iterate through a list (call the iterator sample) and print each element. Also demonstrate range.

my_list = ["course", "coffee", "please"]
for sample in my_list:
    print(sample)
````

````python
print(sample)
````

Read this as:

> For each `sample` in the list `samples`, print the sample.

The variable `sample` temporarily represents one item from the list at a time.


````python
# iterate through a list, create a variable that adds 5 to the term and print that variable

my_list = [2, 4, 6, 8]
for sample in my_list:
    var = sample + 5
    print(var)
````

We can combine loops with f-strings:

````python
# demonstrate iterating through a list of strings and printing a message that includes the string
for x in my_list:
    print(f"Value is {x}")
````

And we can combine loops with `if` statements:

````python
# demonstrate the use of a for loop to iterate over a list of numbers and print whether they are greater than 20

my_list = [5, 10, 20, 30]
for x in my_list:
    print(x)
    if x > 15:
        print("Greater than 15")
````

We might also want access to the index when iterating through a list. This is done using ```enumerate()```

````python
# demonstrate enumeration
for i,num in enumerate(my_list):
    print(i, num)
````

# 8. Building lists with `.append()`

Sometimes we want to create a new list while looping over another list.

Start with an empty list:


````python
# create and print an empty list
my_list = []
print(my_list, len(my_list))
````

Then add items using `.append()`.

````python
# append some values and print the updated list
for i in range(3,8):
    # print(i)
    my_list.append(i)
    print(my_list)
````

````python
my_list
````

This becomes especially useful inside a loop.

````python
# use a for loop to double each element of a list, append the result to a new list and print this
my_list_double = []
for value in my_list:
    my_list_double.append(value * 2)

print(my_list_double)
````

Let's follow what happens:

1. `doubled_numbers` starts empty
2. Python takes the first number
3. it calculates `number * 2`
4. `.append()` adds the result to the new list
5. the loop repeats for every number


Another example: keep only values that pass a condition.

````python
my_list
````

````python
# append values to a list that pass some condition
my_list_condition = []
for value in my_list:
    if value > 5:
        my_list_condition.append(value)

my_list_condition
````
